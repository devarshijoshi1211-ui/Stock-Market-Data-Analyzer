````python

# ============================================
# PHASE 4: DATA CLEANING
# STEP 1: LOAD ORIGINAL STOCK DATA
# ============================================

import pandas as pd
import os

PROJECT_PATH = "/content/drive/MyDrive/Stock-Market-Data-Analyzer"

raw_path = os.path.join(
    PROJECT_PATH,
    "data/raw/AAPL_historical_data.csv"
        )

        # Read the original CSV with two header rows
df = pd.read_csv(
    raw_path,
    header=[0, 1],
    index_col=0
                )

                    # Remove the extra Date header row
df = df[df.index != "Date"]

                    # Flatten the MultiIndex columns
df.columns = df.columns.get_level_values(0)

                    # Convert index into datetime
df.index = pd.to_datetime(df.index, errors="coerce")

                    # Remove invalid dates
df = df[df.index.notna()]

                    # Set index name
df.index.name = "Date"

print("=" * 55)
print("ORIGINAL STOCK DATA LOADED")
print("=" * 55)

print("\nDataset Preview:")
print(df.head())

print("\nDataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
````

````python

# ============================================
# STEP 2: CLEAN AND VALIDATE STOCK DATA
# ============================================

import numpy as np

print("=" * 55)
print("STOCK DATA CLEANING")
print("=" * 55)

# Convert columns to numeric values
for column in df.columns:
    df[column] = pd.to_numeric(
    df[column],
    errors="coerce"
                        )

                        # Remove duplicate dates
df = df[~df.index.duplicated(keep="first")]

                        # Sort dates
df = df.sort_index()

                        # Replace infinite values
df = df.replace([np.inf, -np.inf], np.nan)

                        # Remove rows with missing values
df = df.dropna()

                        # Validate stock prices
price_columns = ["Close", "High", "Low", "Open"]

                        # Remove records with non-positive prices
df = df[(df[price_columns] > 0).all(axis=1)]

                        # Validate volume
df = df[df["Volume"] >= 0]

                        # Validate OHLC price relationships
df = df[
    (df["High"] >= df[["Open", "Close", "Low"]].max(axis=1))
    &
    (df["Low"] <= df[["Open", "Close", "High"]].min(axis=1))
       ]

print("\nData cleaning completed!")

print("Remaining Records:", len(df))
print("Remaining Columns:", len(df.columns))

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Dates:", df.index.duplicated().sum())

print("\nCleaned Dataset Preview:")
print(df.head())
````

````python

# ============================================
# STEP 3: SAVE CLEANED STOCK DATA
# ============================================

processed_path = os.path.join(
    PROJECT_PATH,
    "data/processed/AAPL_cleaned_data.csv"
        )

df.to_csv(processed_path)

print("=" * 55)
print("CLEANED DATASET SAVED")
print("=" * 55)

print("File Location:")
print(processed_path)

print("\nTotal Records:", len(df))

print("\nSUCCESS: Cleaned dataset saved!")
````

````python

# Calculate daily percentage returns
df["Daily_Return"] = df["Close"].pct_change()

# Calculate rolling 20-day annualized volatility
df["Rolling_20D_Volatility"] = (
    df["Daily_Return"].rolling(window=20).std() * np.sqrt(252)
    )

    # Calculate statistical measures
daily_returns = df["Daily_Return"].dropna()

average_daily_return = daily_returns.mean() * 100
daily_volatility = daily_returns.std()
annualized_volatility = daily_volatility * np.sqrt(252) * 100

best_day = daily_returns.idxmax()
worst_day = daily_returns.idxmin()

best_return = daily_returns.max() * 100
worst_return = daily_returns.min() * 100

cumulative_return = (
        (1 + daily_returns).prod() - 1
        ) * 100

print("\n" + "=" * 55)
print("DAILY RETURNS ANALYSIS")
print("=" * 55)

print(f"Average Daily Return: {average_daily_return:.4f}%")
print(f"Best Trading Day: {best_day.date()}")
print(f"Best Daily Return: {best_return:.4f}%")
print(f"Worst Trading Day: {worst_day.date()}")
print(f"Worst Daily Return: {worst_return:.4f}%")
print(f"Cumulative Return: {cumulative_return:.2f}%")

print("\n" + "=" * 55)
print("VOLATILITY ANALYSIS")
print("=" * 55)

print(f"Daily Volatility: {daily_volatility * 100:.4f}%")
print(f"Annualized Historical Volatility: {annualized_volatility:.2f}%")

print("\nCalculations completed successfully!")
````

````python
import os

PROJECT_PATH = "/content/drive/MyDrive/Stock-Market-Data-Analyzer"

SRC_PATH = os.path.join(PROJECT_PATH, "src")
os.makedirs(SRC_PATH, exist_ok=True)

# Create package initialization file
init_path = os.path.join(SRC_PATH, "__init__.py")

with open(init_path, "w") as file:
    file.write('"""Stock Market Data Analyzer package."""\n')

data_fetcher_code = r'''
import os
import pandas as pd
import yfinance as yf


def fetch_stock_data(ticker, start_date, end_date):
  """
  Download historical stock data from Yahoo Finance.

  Parameters:
    ticker (str): Stock ticker symbol.
    start_date (str): Starting date.
    end_date (str): Ending date (exclusive).

  Returns:
    pandas.DataFrame: Historical stock data.
  """

  try:
    data = yf.download(
        ticker,
        start=start_date,
        end=end_date,
        auto_adjust=True,
        progress=False
    )

    if data.empty:
      raise ValueError(
         f"No stock data found for {ticker}."
      )

    # Handle Yahoo Finance MultiIndex columns
    if isinstance(data.columns, pd.MultiIndex):
      data.columns = data.columns.get_level_values(0)

    data.index.name = "Date"
    return data

  except Exception as error:
    raise RuntimeError(
      f"Failed to download data for {ticker}: {error}"
    ) from error

def load_stock_csv(file_path):
  """
  Load stock data from a local CSV file
  Supports standard CSV files and Yahoo Finance
  CSV files with two header rows.
  """

  if not os.path.exists(file_path):
    raise FileNotFoundError(
      f"CSV file not found: {file_path}"
    )

  # First attempt: standard CSV
  data = pd.read_csv(file_path)

  if "Date" not in data.columns:
    data = pd.read_csv(
      file_path,
      header=[0, 1],
      index_col=0
    )

    data = data[data.index != "Date"]
    data.columns = data.columns.get_level_values(0)
    data.index.name = "Date"
    data = data.reset_index()

  if "Date" not in data.columns:
    raise ValueError(
      "CSV must contain a Date column."
    )

  data["Date"] = pd.to_datetime(
      data["Date"],
      errors="coerce"
  )

  data = data.dropna(subset=["Date"])
  data = data.set_index("Date")
  data.index.name = "Date"

  return data


def save_raw_data(data, ticker, project_path):
  """
  Save downloaded data in the raw data directory.
  """
  raw_path = os.path.join(project_path, "data", "raw")
  os.makedirs(raw_path, exist_ok=True)

  file_path = os.path.join(
      raw_path,
      f"{ticker}_historical_data.csv"
  )

  data.to_csv(file_path)
  return file_path
'''

module_path = os.path.join(SRC_PATH, "data_fetcher.py")

with open(module_path, "w", encoding="utf-8") as file:
    file.write(data_fetcher_code.strip() + "\n")

# Verify created files
print("=" * 55)
print("DATA FETCHER MODULE CREATION")
print("=" * 55)

for path in [init_path, module_path]:
    if os.path.exists(path):
        print("SUCCESS:", os.path.basename(path))
        print("Size:", os.path.getsize(path), "bytes")
    else:
        print("ERROR:", os.path.basename(path))

print("\nData fetcher module created!")
````

````python
import os

PROJECT_PATH = "/content/drive/MyDrive/Stock-Market-Data-Analyzer"

SRC_PATH = os.path.join(PROJECT_PATH, "src")

data_cleaner_code = r'''
import numpy as np
import pandas as pd


def clean_stock_data(data):
    """
    Clean and validate historical stock market data.

    Required columns:
        Open, High, Low, Close, Volume
    """

    if data is None or data.empty:
        raise ValueError("Input stock dataset is empty.")

    df = data.copy()

    # Standardize column names
    df.columns = [
        str(column).strip().title()
        for column in df.columns
    ]

    required_columns = [
        "Open", "High", "Low", "Close", "Volume"
    ]

    missing_columns = [
        column for column in required_columns
        if column not in df.columns
    ]

    if missing_columns:
        raise ValueError(
            f"Missing required columns: {missing_columns}"
        )

    # Convert index to datetime
    df.index = pd.to_datetime(
        df.index,
        errors="coerce"
    )

    df.index.name = "Date"

    # Remove invalid dates and duplicate records
    df = df[df.index.notna()]
    df = df[~df.index.duplicated(keep="last")]
    df = df.sort_index()

    # Convert required fields to numeric
    for column in required_columns:
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

    # Remove infinite values
    df = df.replace(
        [np.inf, -np.inf],
        np.nan
    )

    # Remove rows with missing required values
    df = df.dropna(subset=required_columns)

    # Validate prices
    if (df[["Open", "High", "Low", "Close"]] <= 0).any().any():
        raise ValueError("Stock prices must be positive.")

    if (df["Volume"] < 0).any():
        raise ValueError("Trading volume cannot be negative.")

    # Validate OHLC relationships
    invalid_rows = (
        (df["High"] < df["Low"]) |
        (df["High"] < df["Open"]) |
        (df["High"] < df["Close"]) |
        (df["Low"] > df["Open"]) |
        (df["Low"] > df["Close"])
    )

    if invalid_rows.any():
        raise ValueError(
            f"Found {invalid_rows.sum()} invalid OHLC records."
        )

    if df.empty:
        raise ValueError(
            "No valid records remain after cleaning."
        )

    return df


def save_cleaned_data(data, ticker, project_path):
    """
    Save cleaned stock data to processed directory.
    """
    import os

    processed_path = os.path.join(
        project_path,
        "data",
        "processed"
    )

    os.makedirs(processed_path, exist_ok=True)

    file_path = os.path.join(
        processed_path,
        f"{ticker}_cleaned_data.csv"
    )

    data.to_csv(file_path)

    return file_path
'''

module_path = os.path.join(SRC_PATH, "data_cleaner.py")

with open(module_path, "w", encoding="utf-8") as file:
    file.write(data_cleaner_code.strip() + "\n")

print("SUCCESS: data_cleaner.py created!")
print("File size:", os.path.getsize(module_path), "bytes")
````

````python
import os

PROJECT_PATH = "/content/drive/MyDrive/Stock-Market-Data-Analyzer"
SRC_PATH = os.path.join(PROJECT_PATH, "src")

report_generator_code = r'''
import os
from datetime import datetime


def generate_stock_report(data, summary, ticker, project_path):
    """
    Generate a text report containing stock price, moving-average,
    returns, and volatility analysis.
    """
    if data is None or data.empty:
        raise ValueError("No data available to generate the report.")

    reports_path = os.path.join(project_path, "reports")
    os.makedirs(reports_path, exist_ok=True)

    def get_value(key, default="N/A"):
        value = summary.get(key, default)
        if value is None:
            return default
        return value

    def format_number(key, decimals=2, suffix=""):
        value = get_value(key)
        try:
            return f"{float(value):,.{decimals}f}{suffix}"
        except (TypeError, ValueError):
            return str(value)

    report = f"""
============================================================
              STOCK MARKET DATA ANALYSIS REPORT
============================================================

Stock Ticker: {ticker}
Report Generated: {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}

------------------------------------------------------------
1. DATASET OVERVIEW
------------------------------------------------------------

Total Records: {get_value("total_records")}
Start Date: {get_value("start_date")}
End Date: {get_value("end_date")}

------------------------------------------------------------
2. PRICE ANALYSIS
------------------------------------------------------------

Latest Closing Price: ${format_number("latest_close")}
Highest Closing Price: ${format_number("highest_close")}
Lowest Closing Price: ${format_number("lowest_close")}
Average Closing Price: ${format_number("average_close")}
Average Trading Volume: {format_number("average_volume", 0)}

------------------------------------------------------------
3. MOVING AVERAGE ANALYSIS
------------------------------------------------------------

Latest MA20: ${format_number("latest_ma20")}
Latest MA50: ${format_number("latest_ma50")}
Moving Average Relationship: {get_value("trend")}

------------------------------------------------------------
4. DAILY RETURNS ANALYSIS
------------------------------------------------------------

Average Daily Return: {format_number("average_daily_return", 4, "%")}
Best Trading Day: {get_value("best_day")}
Best Daily Return: {format_number("best_return", 4, "%")}
Worst Trading Day: {get_value("worst_day")}
Worst Daily Return: {format_number("worst_return", 4, "%")}
Cumulative Return: {format_number("cumulative_return", 2, "%")}

------------------------------------------------------------
5. VOLATILITY ANALYSIS
------------------------------------------------------------

Daily Volatility: {format_number("daily_volatility", 4, "%")}
Annualized Historical Volatility: {format_number("annualized_volatility", 2, "%")}

------------------------------------------------------------
6. ANALYTICAL SUMMARY
------------------------------------------------------------

The analysis covers {get_value("total_records")} trading records
for {ticker} between {get_value("start_date")} and
{get_value("end_date")}.

The moving-average relationship is:
{get_value("trend")}

The calculated cumulative return is
{format_number("cumulative_return", 2, "%")}.

Annualized historical volatility is
{format_number("annualized_volatility", 2, "%")}.

These measurements describe historical market behaviour.
They should not be treated as predictions of future performance.

------------------------------------------------------------
7. DISCLAIMER
------------------------------------------------------------

This report is for educational and informational purposes only.
It does not constitute investment, financial, or trading advice.

Historical performance does not guarantee future results.
Always conduct independent research before making financial
decisions.

============================================================
                            END OF REPORT
============================================================
"""

    report_file = os.path.join(
        reports_path,
        f"{ticker}_stock_analysis_report.txt"
    )

    with open(report_file, "w", encoding="utf-8") as file:
        file.write(report.strip() + "\n")

    return report_file
'''

module_path = os.path.join(SRC_PATH, "report_generator.py")

with open(module_path, "w", encoding="utf-8") as file:
    file.write(report_generator_code.strip() + "\n")

print("SUCCESS: report_generator.py created!")
print("File size:", os.path.getsize(module_path), "bytes")
````

````python
import os

PROJECT_PATH = "/content/drive/MyDrive/Stock-Market-Data-Analyzer"

architecture_content = r"""# System Architecture

## 1. Overview

The Stock Market Data Analyzer is a modular Python application that retrieves historical stock market data, validates and cleans the dataset, performs financial analysis, generates visualizations, and produces automated reports.

## 2. Application Workflow

```text
User Input
    |
    v
    Data Fetcher
        |
        v
        Data Cleaner
            |
            v
            Stock Analyzer
                |
                +--------------------+
                |                    |
                v                    v
                Visualization       Report Generator
                    |                    |
                    v                    v
                    Chart Outputs       Text Report
                        |
                        v
                        Saved Project Outputs
```

## 3. Module Responsibilities

### data_fetcher.py
- Retrieves historical data through Yahoo Finance.
- Loads stock data from CSV files.
- Saves downloaded raw data.

### data_cleaner.py
- Standardizes column names.
- Converts dates and numerical values.
- Handles missing and invalid values.
- Removes duplicate records.
- Validates OHLC price relationships and trading volume.

### stock_analyzer.py
- Calculates 20-day moving averages.
- Calculates 50-day moving averages.
- Calculates daily returns.
- Calculates rolling 20-day volatility.
- Produces a financial summary.

### visualization.py
- Creates closing price charts.
- Creates trading volume charts.
- Visualizes moving averages.
- Visualizes daily returns and their distribution.
- Visualizes rolling volatility.
- Generates a combined analysis dashboard.

### report_generator.py
- Generates a structured text report.
- Includes dataset statistics, price analysis, returns, moving averages, and volatility.
- Adds an educational disclaimer.

### main.py
- Coordinates the complete application.
- Accepts user inputs.
- Connects the individual modules.
- Saves the final outputs.

## 4. Data Flow

1. The user supplies a ticker and date range.
2. Historical data is downloaded or loaded from CSV.
3. The cleaner validates and prepares the dataset.
4. The analyzer calculates financial indicators.
5. The visualization module generates charts.
6. The report generator creates a text report.
7. All outputs are saved in the project directories.

## 5. Technologies

- Python: application development
- Pandas: data manipulation
- NumPy: numerical calculations
- yfinance: historical stock data
- Matplotlib: chart generation
- Seaborn: statistical visualization

## 6. Error Handling and Data Validation

The application checks for empty datasets, validates required financial fields, removes invalid dates and duplicate records, and checks price and volume constraints.

If online retrieval fails, the application can use an existing ticker-specific CSV file when available.

## 7. Limitations

- Results depend on the availability and quality of historical market data.
- Historical indicators do not predict future market performance.
- The current version generates static visualizations and text reports.
- The application is an educational analysis tool, not an investment advisory system.
"""

architecture_path = os.path.join(PROJECT_PATH, "docs", "architecture.md")

os.makedirs(os.path.dirname(architecture_path), exist_ok=True)

with open(architecture_path, "w", encoding="utf-8") as file:
    file.write(architecture_content)

print("SUCCESS: architecture.md created!")
print("File size:", os.path.getsize(architecture_path), "bytes")
````
